```python
import pandas as pd
import os
import sys
import numpy as np
from bs4 import BeautifulSoup
from urllib.request import urlopen
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from typing import Optional

sys.path.append(os.path.dirname(os.getcwd()))
from utils import Constantes, SimpleLogger

logger = SimpleLogger(f"{Constantes.ruta_log.value}/app.log", "web_scraping")
```

```python
def web_scrapper(lk: str) -> Optional[BeautifulSoup]:
    if not isinstance(lk, str):
        logger.error(f"web_scrapper: {lk} no es str, sino {type(lk)}")
        return None
    try:
        # Configurar proceso de web scraping
        options = Options()
        options.add_argument("--headless")  # No abrir ventana
        options.add_argument("--disable-gpu")
        options.add_argument("--no-sandbox")
        # Objeto tipo webdriver para el link
        tag_browser = webdriver.Chrome(options=options)
        # Usar el webdriver para un request al link
        tag_browser.get(lk)
        # Obtener el html del link una vez cargado
        tag_page_html = tag_browser.page_source
        tag_browser.quit()
        # Parsear el HTML con BeautifulSoup
        tag_page_soup = BeautifulSoup(tag_page_html, "html.parser")
        return tag_page_soup
    except Exception as e:
        logger.exception(f"web_scrapper: {e}")
        return None


def procesar_datos(lk: str, tag_titles: list[str], tag_ws: list[str]) -> Optional[dict]:
    if not isinstance(lk, str):
        logger.error(f"procesar_datos: {lk} no es str, sino {type(lk)}")
        return None
    if len(tag_titles) != len(tag_ws):
        logger.error("procesar_datos: tag_titles y tag_ws tienen diferentes longitudes")
        return None
    # Obtener la información del web_scrapper
    tag_page_soup = web_scrapper(lk)
    if tag_page_soup is None:
        return None
    # Generar una lista por cada columna de la tabla
    res_ws = {}
    for col, ws_col in zip(tag_titles, tag_ws):
        res_ws[col] = tag_page_soup.find_all("td", {"class": ws_col})
    return res_ws


def extraer_scores(lk_scores: str) -> Optional[dict]:
    if not isinstance(lk_scores, str):
        logger.error(f"extraer_scores: {lk_scores} no es str, sino {type(lk_scores)}")
        return None
    # Obtener la información del web_scrapper
    # Generar una lista por cada columna de la tabla
    res_ws = {}
    for col, ws_col in zip(Constantes.scores_titles.value, Constantes.scores_ws.value):
        res_ws[col] = scores_page_soup.findAll("td", {"class": ws_col})
    return res_ws


def extraer_datos_stats(lk_stats: str) -> Optional[dict]:
    if not isinstance(lk_stats, str):
        return None
    # Generar una lista por cada columna de la tabla
    res_ws = {}
    for col, ws_col in zip(Constantes.stats_titles.value, Constantes.stats_ws.value):
        res_ws[col] = stats_page_soup.findAll("td", {"class": ws_col})
    return res_ws
```

```python
(
    full_address_list,
    streetAddress_list,
    addressLocality_list,
    addressRegion_list,
    postalCode_list,
    addressCountry_list,
) = [], [], [], [], [], []
for web in link_universidades:
    if web != "":
        page = urlopen(web)
        page_html = soup(page, "html.parser")

        full_address = page_html.findAll(
            "div",
            {
                "class": "institution-info__contact-detail institution-info__contact-detail--address"
            },
        )[0].text.strip()
        full_address_list.append(full_address)

        if len(full_address.split(", ")) == 3:
            streetAddress_list.append("Sin dirección")
            addressLocality_list.append(full_address.split(", ")[0])
            addressRegion_list.append(full_address.split(", ")[1])
            postalCode_list.append("Sin código postal")
            addressCountry_list.append(full_address.split(", ")[2])
            print(
                f"{len(full_address_list)} out of {len(link_universidades)} - {full_address}"
            )
        elif len(full_address.split(", ")) == 4:
            streetAddress_list.append("Sin dirección")
            addressLocality_list.append(full_address.split(", ")[0])
            addressRegion_list.append(full_address.split(", ")[1])
            postalCode_list.append(full_address.split(", ")[2])
            addressCountry_list.append(full_address.split(", ")[3])
            print(
                f"{len(full_address_list)} out of {len(link_universidades)} - {full_address}"
            )
        elif len(full_address.split(", ")) == 5:
            streetAddress_list.append(full_address.split(", ")[0])
            addressLocality_list.append(full_address.split(", ")[1])
            addressRegion_list.append(full_address.split(", ")[2])
            postalCode_list.append(full_address.split(", ")[3])
            addressCountry_list.append(full_address.split(", ")[4])
            print(
                f"{len(full_address_list)} out of {len(link_universidades)} - {full_address}"
            )
        elif len(full_address.split(", ")) == 6:
            streetAddress_list.append(", ".join(full_address.split(", ")[0:2]))
            addressLocality_list.append(full_address.split(", ")[2])
            addressRegion_list.append(full_address.split(", ")[3])
            postalCode_list.append(full_address.split(", ")[4])
            addressCountry_list.append(full_address.split(", ")[5])
            print(
                f"{len(full_address_list)} out of {len(link_universidades)} - {full_address}"
            )
        else:
            print(
                f"{len(full_address_list)} out of {len(link_universidades)} - Problemas - {full_address} - {len(full_address.split(', '))}"
            )
    else:
        full_address_list.append("")
        streetAddress_list.append("")
        addressLocality_list.append("")
        addressRegion_list.append("")
        postalCode_list.append("")
        addressCountry_list.append("")
        print(f"{len(full_address_list)} out of {len(link_universidades)} - Sin link")
```

```python
df = pd.DataFrame(
    {
        "rank": ranking,
        "name": nombres_universidad,
        "number_students": cantidad_estudiantes,
        "student_staff_ratio": proporcion_estudiantes_staff,
        "intl_students": cantidad_estudiantes_internacionales,
        "female_male_ratio": ratio_mujeres_hombres,
        "%_interdisc_science_research": porcentaje_investigacion_interdisciplinaria,
        "overall_score": overall_score,
        "teaching_score": teaching_score,
        "research_score": research_score,
        "citations_score": citations_score,
        "industry_income_score": industry_income_score,
        "international_outlook_score": international_outlook_score,
        "address": full_address_list,
        "street_address": streetAddress_list,
        "locality_address": addressLocality_list,
        "region_address": addressRegion_list,
        "postcode_address": postalCode_list,
        "country_address": addressCountry_list,
    }
)

df["number_students"] = df["number_students"].str.replace(pat=",", repl="", regex=True)
df = df.replace("n/a*", np.nan, regex=True)
```
